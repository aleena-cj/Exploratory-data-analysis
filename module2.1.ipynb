```python
# 2
import pandas as pd

population = pd.Series(
    [20.96, 32.94, 13.14, 11.50, 15.22],
    index=['Mumbai', 'Delhi', 'Bengaluru', 'Chennai', 'Kolkata']
)

print(population['Bengaluru'])
print(population['Delhi':'Chennai'])
```

```python
# 3
import pandas as pd

products = {
    'Laptop': 55000,
    'Phone': 25000,
    'Tablet': 18000
}

prices = pd.Series(products)

print(prices)
print(prices.index)
print(prices.index.equals(pd.Index(products.keys())))
```

```python
# 4
import pandas as pd

area = pd.Series(
    [307713, 342239, 191791, 130060, 196024],
    index=['Rajasthan', 'Madhya Pradesh', 'Maharashtra', 'Tamil Nadu', 'Karnataka']
)

population = pd.Series(
    [81.0, 72.6, 126.4, 76.9, 68.0],
    index=['Rajasthan', 'Madhya Pradesh', 'Maharashtra', 'Tamil Nadu', 'Karnataka']
)

states = pd.DataFrame({
    'area': area,
    'population': population
})

states['density'] = states['population'] / states['area']

print(states)
```
