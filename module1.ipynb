```
# 14
import numpy as np

np.random.seed(0)

a = np.random.randint(0, 10, size=(2, 3))
b = np.random.randint(0, 10, size=(2, 3))

vertical1 = np.vstack((a, b))
vertical2 = np.concatenate((a, b), axis=0)

horizontal1 = np.hstack((a, b))
horizontal2 = np.concatenate((a, b), axis=1)

print(a)
print(b)
print(vertical1)
print(vertical2)
print(horizontal1)
print(horizontal2)

print(np.array_equal(vertical1, vertical2))
print(np.array_equal(horizontal1, horizontal2))
```

```
# 22
import numpy as np

np.random.seed(0)

X = np.random.rand(10, 2)

dist_sq = np.sum(
    (X[:, np.newaxis, :] - X[np.newaxis, :, :]) ** 2,
    axis=-1
)

nearest = np.argsort(dist_sq, axis=1)[:, 1:3]

print(X)
print(nearest)
```

```
# 24
import numpy as np

dtype = [
    ('title', 'U30'),
    ('author', 'U20'),
    ('price', 'f4')
]

books = np.array([
    ('Python Basics', 'John Smith', 450.0),
    ('Data Science', 'Jane Doe', 550.0),
    ('Machine Learning', 'Alex Brown', 650.0),
    ('Deep Learning', 'Mary Jones', 750.0)
], dtype=dtype)

print(books['title'])
print(books[1])

amount = 600
print(books[books['price'] < amount]['title'])
```

```
# 27
import numpy as np

dtype = [
    ('roll_no', 'i4'),
    ('name', 'U20'),
    ('marks', 'f4')
]

students = np.array([
    (1, 'Anu', 85.5),
    (2, 'Binu', 72.0),
    (3, 'Cathy', 91.5),
    (4, 'Diya', 68.0),
    (5, 'Eva', 88.0)
], dtype=dtype)

sorted_students = np.sort(students, order='marks')[::-1]

print(sorted_students)
```
